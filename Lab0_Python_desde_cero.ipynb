## 6. Listas y bucles `for`: trabajar con muchos valores

📘 Una **lista** guarda varios valores en orden, entre corchetes `[ ]`. Cada elemento tiene una **posición (índice)** que **empieza en 0**.

````python
pris_us = [1000, 1150, 1300, 1000, 1150, 1300]    # PRI de pulsos consecutivos (¿te suena a stagger?)

print(pris_us[0])        # primer elemento (índice 0)
print(pris_us[2])        # tercer elemento (índice 2)
print(pris_us[-1])       # el último
print(pris_us[0:3])      # del índice 0 al 2 (el 3 NO se incluye)
print(len(pris_us))      # cuántos elementos hay
````

````python
print("mínimo:", min(pris_us), " máximo:", max(pris_us), " suma:", sum(pris_us))
print("promedio:", sum(pris_us) / len(pris_us))
pris_us.append(1000)     # agregar un elemento al final
print(pris_us)
````

📘 Un bucle **`for`** repite instrucciones **para cada elemento** de una lista. Lo que se repite va **indentado** (4 espacios a la derecha).

````python
for pri in pris_us:
    prf = 1e6 / pri                      # PRI en µs → PRF en Hz
    print(f"PRI = {pri:>5} µs   →   PRF = {prf:7.1f} Hz")
print("Fin del bucle")                    # esta línea NO está indentada: se ejecuta una sola vez
````

⚠️ **La indentación es parte del lenguaje.** Si olvidas los 4 espacios, Python da `IndentationError`.

## 8. Funciones: tus propias herramientas de cálculo

📘 Una **función** empaqueta un cálculo para reutilizarlo. Se define con `def`, recibe **parámetros** y devuelve un resultado con `return`. El texto entre `""" """` (*docstring*) documenta qué hace.

```python
def nombre_funcion(parametro1, parametro2):
    """Qué hace la función."""
    resultado = ...           # cálculos
    return resultado
```

A continuación construiremos el **kit de fórmulas radar** que usarás durante todo el curso.

| Magnitud | Fórmula | ¿Qué nos dice? |
|---|---|---|
| Longitud de onda | $\lambda = c / f$ | Tamaño de antena, Doppler |
| PRF | $PRF = 1 / PRI$ | Pulsos por segundo |
| Alcance no ambiguo | $R_{ua} = c \cdot PRI / 2$ | Distancia máxima sin ambigüedad → **función del radar** |
| Resolución en distancia | $\Delta R = c \cdot PW / 2$ | Separación mínima entre blancos (pulso sin modulación) |
| Ciclo de trabajo | $d = PW / PRI$ | Fracción del tiempo transmitiendo |
| Desplazamiento Doppler | $f_D = 2 v / \lambda$ | Velocidad radial del blanco |

````python
C_LUZ = 299_792_458.0    # m/s

def longitud_onda_m(f_mhz):
    """Longitud de onda en metros a partir de la frecuencia en MHz."""
    return C_LUZ / (f_mhz * 1e6)

def prf_hz(pri_us):
    """PRF en Hz a partir de la PRI en microsegundos."""
    return 1e6 / pri_us

def alcance_no_ambiguo_km(pri_us):
    """Alcance máximo no ambiguo (km): R = c·PRI/2."""
    return C_LUZ * (pri_us * 1e-6) / 2 / 1000

def resolucion_distancia_m(pw_us):
    """Resolución en distancia (m) de un pulso sin modulación: ΔR = c·PW/2."""
    return C_LUZ * (pw_us * 1e-6) / 2

def ciclo_trabajo_pct(pw_us, pri_us):
    """Ciclo de trabajo en %: PW/PRI."""
    return 100 * pw_us / pri_us

def doppler_hz(v_ms, f_mhz):
    """Desplazamiento Doppler (Hz) de un blanco con velocidad radial v (m/s)."""
    return 2 * v_ms / longitud_onda_m(f_mhz)

print("Funciones definidas ✔")
````

▶️ Ahora **usemos** las funciones (se *llaman* escribiendo su nombre y los valores entre paréntesis):

````python
print(f"λ a 9400 MHz              : {longitud_onda_m(9400)*100:.2f} cm")
print(f"PRF con PRI = 1000 µs     : {prf_hz(1000):.0f} Hz")
print(f"Alcance no ambiguo        : {alcance_no_ambiguo_km(1000):.1f} km")
print(f"Resolución con PW = 1 µs  : {resolucion_distancia_m(1):.0f} m")
print(f"Ciclo de trabajo          : {ciclo_trabajo_pct(1, 1000):.2f} %")
print(f"Doppler de un avión a 250 m/s en banda X : {doppler_hz(250, 9400)/1000:.1f} kHz")
````

🎯 **Interpretación EW.** El alcance no ambiguo es una pista directa de la **función** del radar:

* PRI larga (≥ 1 ms → R<sub>ua</sub> ≥ 150 km): radares de **vigilancia / alerta temprana**.
* PRI corta (≈ 100 µs → 15 km) o PRF alta (> 10 kHz): radares de **seguimiento / control de tiro / pulso-Doppler**, que priorizan la medición de velocidad.

⚠️ **Sobre el Doppler:** la fórmula usa λ en **metros**. En banda X, λ ≈ 0,032 m. Si por error se usa λ = 30 (m), el resultado sale 1000 veces más chico.

### 8.1 Una función con decisiones: banda OTAN / EW

Las bandas que usan los sistemas EW (designación **OTAN**) son distintas de las IEEE:

| OTAN | A | B | C | D | E | F | G | H | I | J | K | L | M |
|---|---|---|---|---|---|---|---|---|---|---|---|---|---|
| GHz | 0–0,25 | 0,25–0,5 | 0,5–1 | 1–2 | 2–3 | 3–4 | 4–6 | 6–8 | 8–10 | 10–20 | 20–40 | 40–60 | 60–100 |

````python
def banda_otan(f_mhz):
    """Devuelve la letra de banda OTAN (EW) para una frecuencia en MHz."""
    f = f_mhz / 1000          # MHz → GHz
    if f < 0.25:  return "A"
    elif f < 0.5: return "B"
    elif f < 1:   return "C"
    elif f < 2:   return "D"
    elif f < 3:   return "E"
    elif f < 4:   return "F"
    elif f < 6:   return "G"
    elif f < 8:   return "H"
    elif f < 10:  return "I"
    elif f < 20:  return "J"
    elif f < 40:  return "K"
    elif f < 60:  return "L"
    else:         return "M"

for f in [1300, 2950, 5650, 9400, 15600]:
    print(f"{f:>6} MHz  →  banda OTAN {banda_otan(f)}")
````

### 8.2 Desafío EW: relación J/S (Jamming / Señal)

📘 Para un enlace de **un solo tramo** (p. ej. interferencia a un receptor distante; formulación de D. Adamy):

$$\frac{J}{S}\,(dB) = P_J - P_T + G_J - G_T - 20\log_{10}(D_J) + 20\log_{10}(D_S) + G_{RJ} - G_R$$

| Símbolo | Significado | Unidad |
|---|---|---|
| $P_J$, $P_T$ | Potencia del *jammer* y del transmisor deseado | dBm |
| $G_J$, $G_T$ | Ganancia de antena del *jammer* y del transmisor | dB |
| $D_J$, $D_S$ | Distancia *jammer*→receptor y transmisor→receptor | km |
| $G_{RJ}$ | Ganancia de la antena receptora **hacia el jammer** | dB |
| $G_R$ | Ganancia de la antena receptora **hacia el transmisor** | dB |

⚠️ Observa el **signo de $G_R$: es negativo**. Más ganancia hacia la señal deseada *reduce* J/S.

````python
def js_db(Pj, Pt, Gj, Gt, Dj_km, Ds_km, Grj, Gr):
    """Relación J/S en dB para un enlace de un tramo."""
    return Pj - Pt + Gj - Gt - 20 * math.log10(Dj_km) + 20 * math.log10(Ds_km) + Grj - Gr

# Jammer de 100 W (50 dBm) a 50 km; transmisor de 10 W (40 dBm) a 20 km
js = js_db(Pj=50, Pt=40, Gj=10, Gt=3, Dj_km=50, Ds_km=20, Grj=0, Gr=3)
print(f"J/S = {js:.1f} dB  →  {'la interferencia domina' if js > 0 else 'la señal domina'}")
````

💡 Al llamar la función escribimos `Pj=50, Pt=40, ...` (**argumentos con nombre**). Así es imposible confundir el orden de 8 parámetros.

## 9. Diccionarios: la "ficha" de un emisor

📘 Un **diccionario** guarda pares **clave → valor** entre llaves `{ }`. Es ideal para describir un emisor con sus parámetros.

````python
emisor = {
    "id": "E-01",
    "freq_mhz": 2950,
    "pri_us": 1000,
    "pw_us": 1.2,
    "scan": "Circular",
    "periodo_scan_s": 6.0,
}

print(emisor["freq_mhz"])          # leer un valor por su clave
emisor["aoa_deg"] = 135            # agregar una clave nueva
print(emisor)
````

🎯 Combinemos todo: con el diccionario y tus funciones, genera una **mini-ficha técnica** automática.

````python
def mini_ficha(e):
    """Imprime una ficha técnica con parámetros medidos y derivados."""
    print("=" * 52)
    print(f" FICHA TÉCNICA PRELIMINAR · Emisor {e['id']}")
    print("=" * 52)
    print(f" RF ................ {e['freq_mhz']} MHz  (banda OTAN {banda_otan(e['freq_mhz'])})")
    print(f" λ ................. {longitud_onda_m(e['freq_mhz'])*100:.1f} cm")
    print(f" PRI / PRF ......... {e['pri_us']} µs / {prf_hz(e['pri_us']):.0f} Hz")
    print(f" PW ................ {e['pw_us']} µs")
    print(f" Alcance no ambiguo  {alcance_no_ambiguo_km(e['pri_us']):.0f} km")
    print(f" Resolución dist. .. {resolucion_distancia_m(e['pw_us']):.0f} m")
    print(f" Ciclo de trabajo .. {ciclo_trabajo_pct(e['pw_us'], e['pri_us']):.2f} %")
    if e.get("periodo_scan_s"):
        print(f" Scan .............. {e['scan']} · {e['periodo_scan_s']} s ({60/e['periodo_scan_s']:.0f} rpm)")
    else:
        print(f" Scan .............. {e['scan']} (sin periodo: iluminación continua)")
    print(f" AOA ............... {e.get('aoa_deg', 's/d')}°")
    print("=" * 52)

mini_ficha(emisor)
````

## 10. NumPy: cálculos con miles de pulsos a la vez

📘 Un receptor ESM entrega **miles de pulsos**. Recorrerlos uno a uno con `for` es lento. **NumPy** permite operar sobre **arreglos** (*arrays*) completos con una sola instrucción (cálculo **vectorial**).

Por convención se importa así: `import numpy as np`

````python
import numpy as np

pris = np.array([1000, 1150, 1300, 1000, 1150, 1300])   # µs

prfs = 1e6 / pris                    # ¡se aplica a TODOS los elementos!
rangos_km = C_LUZ * pris * 1e-6 / 2 / 1000

print("PRF (Hz):", np.round(prfs, 1))
print("Rua (km):", np.round(rangos_km, 1))
print(f"Media: {np.mean(pris):.1f}   Mediana: {np.median(pris):.1f}   Desv. estándar: {np.std(pris):.1f}")
````

💡 Funciones de NumPy que usarás mucho:

| Función | Para qué |
|---|---|
| `np.mean`, `np.median`, `np.std` | media, mediana, desviación estándar |
| `np.min`, `np.max` | mínimo, máximo |
| `np.diff(x)` | diferencias entre elementos consecutivos → **¡así se calcula la PRI a partir del TOA!** |
| `np.linspace(a, b, n)` | `n` valores equiespaciados entre `a` y `b` |
| `np.round(x, d)` | redondear a `d` decimales |

🎯 Un receptor ESM mide el **TOA** (*Time Of Arrival*) de cada pulso. La **PRI** se obtiene restando TOA consecutivos:

````python
toa_us = np.array([0, 1000, 2150, 3450, 4450, 5600, 6900])   # tiempos de llegada medidos
dtoa_us = np.diff(toa_us)                                    # diferencias consecutivas
print("DTOA (µs):", dtoa_us)
````

🎯 Esa secuencia `1000 → 1150 → 1300 → 1000 → …` es un **stagger de 3 posiciones**. Lo estudiaremos a fondo en el Lab 3.

### 11.1 Visualizar un tren de pulsos

🎯 Dibujemos lo que transmite un radar: pulsos de ancho **PW** que se repiten cada **PRI**. (Exageramos el PW para que se vea.)

````python
pri = 100        # µs
pw = 20          # µs (exagerado para visualizar)
t = np.linspace(0, 450, 4000)                  # eje de tiempo en µs
tren = np.where((t % pri) < pw, 1.0, 0.0)      # 1 si estamos "dentro" de un pulso, 0 si no

plt.figure(figsize=(11, 3))
plt.plot(t, tren, lw=2)
plt.annotate("", xy=(100, 1.15), xytext=(0, 1.15), arrowprops=dict(arrowstyle="<->"))
plt.text(50, 1.2, "PRI", ha="center")
plt.annotate("", xy=(20, 0.5), xytext=(0, 0.5), arrowprops=dict(arrowstyle="<->", color="red"))
plt.text(10, 0.57, "PW", ha="center", color="red")
plt.ylim(-0.1, 1.4)
plt.xlabel("Tiempo (µs)"); plt.ylabel("Amplitud (normalizada)")
plt.title("Tren de pulsos: PW y PRI")
plt.show()
````
